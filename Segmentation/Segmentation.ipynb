```python
from PIL import Image
import matplotlib.pyplot as plt
import requests
from io import BytesIO
from ultralytics import YOLO

# Load the image from the URL

#url = "https://www.unibz.it/assets/Images/1-University/Hero-Home/2024-12-NOI-Techpark-Campus-9907-copy__FocusFillMaxWyIwLjE4IiwiMC4yNSIsMjQ1MCwxODM4XQ.jpg"
url = "https://www.unibz.it/assets/Images/Faculty-Scitech/2024-04-05-Master-in-Environmental-Management-of-Mountain-Areas__FocusFillMaxWyIwLjAwIiwiMC4wMCIsMjQ1MCwxODM4XQ.jpg"
response = requests.get(url)
img = Image.open(BytesIO(response.content))

# Load a pretrained YOLO model
model = YOLO("yolo11n-seg.pt")

# Run inference on an image
results = model(url)  # results list

# Plot the original image and the result with masks
plt.figure(figsize=(12, 6))

# Original Image
plt.subplot(1, 2, 1)
plt.title("Original Image")
plt.imshow(img)
plt.axis('off')

# Image with YOLO Segmentation
plt.subplot(1, 2, 2)
plt.title("YOLO Segmentation Result")
plt.imshow(results[0].plot()[:, :, ::-1])
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
from PIL import Image
import os

def convert_jpg_to_png(folder_path):
    # Make sure the folder exists
    if not os.path.isdir(folder_path):
        print(f"Folder not found: {folder_path}")
        return
    
    # Iterate through all files in the folder
    for filename in os.listdir(folder_path):
        if filename.lower().endswith(".jpg") or filename.lower().endswith(".jpeg"):
            jpg_path = os.path.join(folder_path, filename)
            png_filename = os.path.splitext(filename)[0] + ".png"
            png_path = os.path.join(folder_path, png_filename)
            
            # Open JPG and save as PNG
            with Image.open(jpg_path) as img:
                img.save(png_path, "PNG")
            
            print(f"Converted: {filename} -> {png_filename}")

# Example usage:
folder = "/home/tb/knime-workspace/Example Workflows/ImageProcessing/04_Car_Counting/data"
convert_jpg_to_png(folder)
```
